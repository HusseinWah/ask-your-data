```python
df = df.drop_duplicates()
df.shape
```

```python
df["region"] = df["region"].str.strip().str.title().replace({"Alex": "Alexandria"})
df["region"].unique()
```

```python
df["revenue"] = df["revenue"].str.replace("EGP", "").str.replace(",", "").str.strip().astype(int)
df["revenue"].head()
```

```python
iso = pd.to_datetime(df["order_date"], format="%Y-%m-%d", errors="coerce")
dmy = pd.to_datetime(df["order_date"], format="%d/%m/%Y", errors="coerce")
df["order_date"] = iso.fillna(dmy)
df["order_date"].isna().sum()
```

```python
prices = {"Desk Lamp": 450, "Office Chair": 2800, "Bookshelf": 1900, "Side Table": 1200, "Floor Lamp": 950}
df["units_sold"] = (df["revenue"] / df["product"].map(prices)).astype(int)
df["units_sold"].describe()
```

```python
df["sales_rep"] = df["sales_rep"].fillna("Unknown")
```

```python
print(df.isna().sum().sum(), "empty cells left")
print(df.dtypes)
df.to_csv("clean_sales.csv", index=False)
```
